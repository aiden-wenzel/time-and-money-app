```python
# Imports
import pandas as pd
import numpy as np
import datetime as dt
```

```python
# Read in the spending data
data = pd.read_csv("./experimental.csv")
data
```

```python
# Sum the total amount spent
amounts = data["amount"]
np.sum(amounts.to_numpy())
```

```python
# Datetime
test_date_str = data["date"][0]
test_date = dt.datetime.strptime(test_date_str, "%Y/%m/%d")
test_date.year
```

```python
# Convert date strings in original dataframe into datetimes

dates_timestamp = []
for date in data["date"]:
    dates_timestamp.append(pd.Timestamp(date))

data["date"] = pd.Series(dates_timestamp)
data
```
